4 Задание


```python
tracks_after_2024 = df[df['release_year'] > 2024]

print(f"Количество треков, выпущенных после 2024 года: {len(tracks_after_2024)}")
display(tracks_after_2024.head())
```

```python
mean_pop = df['popularity'].mean()
median_pop = df['popularity'].median()

print(f"Среднее значение popularity: {mean_pop:.2f}")
print(f"Медиана popularity: {median_pop:.2f}")

# Минимум, максимум и стандартное отклонение для popularity
min_pop = df['popularity'].min()
max_pop = df['popularity'].max()
std_pop = df['popularity'].std()

print(f"Минимум popularity: {min_pop}")
print(f"Максимум popularity: {max_pop}")
print(f"Стандартное отклонение popularity: {std_pop:.2f}")
```

```python
print("Описательные статистики для popularity:")
print(df['popularity'].describe())
```

В этой таблице:
count — количество значений
mean — среднее
std — стандартное отклонение
min — минимум
25% — первый квартиль
50% — медиана
75% — третий квартиль
max — максимум
